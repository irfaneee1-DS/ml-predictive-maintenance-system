```python
# ============================================
# Create Final 23-Feature Random Forest Model
# ============================================

import pandas as pd
import joblib
from pathlib import Path
from sklearn.ensemble import RandomForestClassifier


# 1. Load the final engineered dataset
df = pd.read_csv("phase1_full_dataset.csv")


# 2. Define the exact 23 predictor columns
feature_columns = [
    "Mean",
    "RMS",
    "Std",
    "Peak",
    "Peak_to_Peak",
    "Skewness",
    "Kurtosis",
    "Crest_Factor",
    "Shape_Factor",
    "Impulse_Factor",
    "Spectral_Energy",
    "Dominant_Frequency",
    "Amp_BPFO",
    "Amp_BPFI",
    "Amp_BSF",
    "BPFO_BPFI_Ratio",
    "Band_0_500",
    "Band_500_1000",
    "Band_1000_2000",
    "Wavelet_D1",
    "Wavelet_D2",
    "Wavelet_D3",
    "Wavelet_D4"
]


# 3. Separate predictors and target
X = df[feature_columns]
y = df["Label"]


# 4. Create the Random Forest model
rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)


# 5. Train the deployment model
rf_model.fit(X, y)


# 6. Create the models folder if needed
models_dir = Path("models")
models_dir.mkdir(exist_ok=True)


# 7. Save the trained model
model_path = models_dir / "random_forest_23_features.pkl"
joblib.dump(rf_model, model_path)


# 8. Save the exact feature order
feature_path = models_dir / "random_forest_23_features_columns.txt"

with open(feature_path, "w") as f:
    for feature in feature_columns:
        f.write(feature + "\n")


print("Model saved:", model_path)
print("Feature schema saved:", feature_path)
print("Number of predictors:", len(feature_columns))
print("Training rows:", len(X))
```
